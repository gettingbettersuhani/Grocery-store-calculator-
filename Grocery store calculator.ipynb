# GROCERY STORE CALCULATOR - CASE STUDY 1

```python
print("Hello, Welcome to Save Mart")
name = input("Please Enter your Name: ")
contact = int(input("Please Enter your Contact Number: "))

a = int(input("Enter Number of items purchased: "))
items_dict = {}
total_sum = 0  # Use total_sum instead of sum to avoid conflicts with the built-in sum function
for i in range(a):
    b = input("Enter item purchased: ")
    c = int(input("Enter Price of item: "))  
    d = int(input("Enter Number of items purchased: "))
    items_dict[b] = c * d
    total_sum += c * d  # Correct summation

# Function to display and edit dictionary
def display_and_edit_dict(items_dict):
    print("Here is the current list of purchased items:")
    print(items_dict)
    
    edit = input("Would you like to edit any item? (yes/no): ")
    while edit.lower() == 'yes':
        item_to_edit = input("Enter the item you want to edit: ")
        if item_to_edit in items_dict:
            new_cost = int(input(f"Enter new Price for {item_to_edit}: "))
            new_quantity = int(input(f"Enter new quantity for {item_to_edit}: "))
            items_dict[item_to_edit] = new_cost * new_quantity
        else:
            print(f"Item {item_to_edit} not found in the list.")
            
        edit = input("Would you like to edit another item? (yes/no): ")
    return items_dict

# Display and edit dictionary
items_dict = display_and_edit_dict(items_dict)

# Function to add more items
def add_more_items(items_dict):
    add = input("Would you like to add more items to your bill? (yes/no): ")
    while add.lower() == 'yes':
        b = input("Enter item purchased: ")
        c = int(input("Enter Price of item: "))  
        d = int(input("Enter Number of items purchased: "))
        if b in items_dict:
            items_dict[b] += c * d
        else:
            items_dict[b] = c * d
        add = input("Would you like to add more items to your bill? (yes/no): ")
    return items_dict

# Add more items if needed
items_dict = add_more_items(items_dict)
total_sum = sum(items_dict.values())  # Correct way to calculate the sum of the values

discount = input("Do you have a discount code? (yes/no): ")
if discount.lower() == 'yes':
    code = input("Enter discount coupon code: ")
    discount_percentage = int(input("Enter discount percentage: "))
    discount_amount = (discount_percentage / 100) * total_sum
    total_sum -= discount_amount
else:
    discount_percentage = 0

print("Here is your final bill")
print("These are the items you have purchased")
print(items_dict)
print(f'{name}, your total purchase amount is {total_sum:.2f} after a {discount_percentage}% discount')
print(f'Thank you, {name}, for shopping. Hope you enjoyed!')
```
